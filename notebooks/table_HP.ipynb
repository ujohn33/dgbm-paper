```python
import pandas as pd
import os
import json

def read_jsons_to_dataframe(folder_path):
    # Initialize an empty list to store dataframes
    dataframes = []
    
    # Loop through each file in the folder
    for filename in os.listdir(folder_path):
        # Check if the file is a JSON file
        if filename.endswith('.json'):
            # Extract dataset name from filename (before the first underscore)
            dataset_name = filename.split('_', 1)[0]
            
            file_path = os.path.join(folder_path, filename)
            
            # Open and load the JSON file
            with open(file_path, 'r') as file:
                data = json.load(file)
            
            # Create a DataFrame from the loaded data
            df = pd.DataFrame([data])

            # Insert the dataset name at the first position
            df.insert(0, 'dataset', dataset_name)
            
            # Append the DataFrame to the list
            dataframes.append(df)
    
    # Concatenate all dataframes into one
    final_dataframe = pd.concat(dataframes, ignore_index=True)
    
    return final_dataframe
```

```python
# Define the folder path where your JSON files are stored
folder_path = 'logs/exp/'

# Call the function and get the dataframe
final_dataframe = read_jsons_to_dataframe(folder_path)

# Save the dataframe to a CSV file
csv_file_path = 'results/HP_exp.csv'
final_dataframe.to_csv(csv_file_path, index=False)

print(f'Data saved to {csv_file_path}')
```
